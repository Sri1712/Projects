### 1. Artificial Neural Networks (ANN)
* An ANN is inspired by the structure of the human brain and the most basic type of neural network. It consists of input, hidden, and output layers made up of neurons.
* Each neuron takes an input, multiplies it by a weight, adds bias, and applies an activation function.
* The network learns by adjusting weights using backpropagation and gradient descent.

**Key Concepts:**
- Activation functions (ReLU, Sigmoid, Tanh)
- Backpropagation + gradient descent

**Example:** <br>
Image classification (basic), <br>
Spam detection, Sentiment analysis <br>
Regression (house price prediction) <br>

### 2. Convolutional Neural Networks (CNN)
* A CNN is designed to work with image data. It can detect visual features like edges, colors, and shapes, making it excellent for vision tasks.
* Early layers learn edges, textures whereas deeper layers learn shapes, objects
* They use filters/kernels to capture spatial hierarchies in pixel data.

**Key Concepts:**
* Convolution Layer: Applies filters to detect features like edges, shapes, textures.
* ReLU Activation: Adds non-linearity so the network can learn complex patterns.
* Pooling Layer: Reduces spatial size (downsampling), keeps important info.
* Fully Connected Layer: Final layer for classification or regression.
* Translation Invariance: Learns patterns regardless of position in the input.

**Example:** <br>
Image classification (cats vs dogs) <br>
Facial recognition <br>
Medical imaging (detecting tumors) <br>
Self-driving car vision <br>

### 3. Recurrent Neural Networks (RNN)
* RNNs are built for sequential data - where previous inputs matter like language or time-series.
* They maintain a hidden state that is updated with each time step.

**Key Concepts:**
* Sequential Processing: Handles inputs one at a time (good for time-based data).
* Hidden State: Stores memory of previous inputs in the sequence.
* Shared Weights: Same weights used at every time step.
* Temporal Dependency: Learns relationships over time (e.g., in language or signals).

**Limitations:**
* Struggles with long sequences (solved by LSTM, GRU).
* LSTM (Long Short-Term Memory) is an advanced type of RNN designed to remember information for long periods and avoid the vanishing gradient problem.
* GRU (Gated Recurrent Unit) is a simplified version of LSTM that performs almost as well, with fewer parameters.

**Example:** <br>
Language modeling <br>
Machine translation <br>
Speech recognition <br>
Stock prediction <br>

### 4. Transformers
* Unlike RNNs, Transformers process sequences in parallel using self-attention.
* Handles long-range dependencies with ease that powers modern LLMs like GPT, BERT, T5

**Key Concepts:**
* Self-Attention: Allows the model to weigh the importance of every token
* Positional Encoding: Injects word order into the model
* Multi-head Attention: Captures diverse contextual relations

**Examples:** <br>
Text summarization <br>
Translation <br>
Chatbots <br>
Coding assistants <br>
