```python
import pandas as pd
import numpy as np
from faker import Faker
from datetime import timedelta
```

```python
fake = Faker()
np.random.seed(42)

num_rows = 150
```

##### *Customers Table*

```python
customers = pd.DataFrame({
    "customer_id": range(1, num_rows + 1),
    "name": [fake.name() for _ in range(num_rows)],
    "email": [fake.email() for _ in range(num_rows)]
})
```

##### *Orders Table*

```python
orders = pd.DataFrame({
    "order_id": range(1001, 1001 + num_rows),
    "customer_id": np.random.choice(customers["customer_id"], num_rows),
    "order_date": [fake.date_between(start_date="-1y", end_date="today") for _ in range(num_rows)],
    "order_amount": np.round(np.random.uniform(20, 500, num_rows), 2),
    "currency": ["USD"] * num_rows
})
```

```python
orders.head()
```

##### *Shipments Table*

```python
status = []
status_choices = ["SHIPPED", "IN_TRANSIT", "DELIVERED"]
```

```python
for order_id, order_date in zip(orders["order_id"], orders["order_date"]):
    shipped_at = order_date + timedelta(days=np.random.randint(1, 5))

    order_status = np.random.choice(status_choices, p=[0.2, 0.3, 0.5])

    delivered_at = (
        shipped_at + timedelta(days=np.random.randint(1, 7))
        if order_status == "DELIVERED"
        else pd.NaT
    )

    status.append({
        "shipment_id": fake.uuid4(),
        "order_id": order_id,
        "status": order_status,
        "shipped_at": shipped_at,
        "delivered_at": delivered_at
    })

shipments = pd.DataFrame(status)
```

```python
shipments.head()
```

```python
customers.to_csv("customers.csv", index=False)
orders.to_csv("orders.csv", index=False)
shipments.to_csv("shipment.csv", index=False)
```
